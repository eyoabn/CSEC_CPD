```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("--- First 5 Rows ---")
print(df.head())

print("\n--- Data Info ---")
print(df.info())

print("\n--- Statistics ---")
print(df.describe())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())


df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df.drop(columns=['Cabin'], inplace=True)


df.drop_duplicates(inplace=True)

print("Data Cleaning Complete!")
```

```python
print("Survival by Gender:")
print(df.groupby('Sex')['Survived'].mean())

print("\nSurvival by Class:")
print(df.groupby('Pclass')['Survived'].mean())

print("\nAverage Age per Class:")
print(df.groupby('Pclass')['Age'].mean())


df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 18, 80], labels=['Child', 'Teen', 'Adult'])
print("\nSurvival by Age Group:")
print(df.groupby('AgeGroup')['Survived'].mean())
```

```python
female_survivors = df[(df['Sex'] == 'female') & (df['Survived'] == 1)]

children_survivors = df[(df['Age'] < 18) & (df['Survived'] == 1)]
first_class = df[df['Pclass'] == 1]

print(f"Number of female survivors: {len(female_survivors)}")
print(f"Number of children survivors: {len(children_survivors)}")
```
